### 🔴 Ejercicio 2.5 — Agrupación y Resumen por Grupo con Diccionarios *(Reto Integrador)*
**Contexto:** En analítica de datos necesitamos agrupar filas por una categoría (como haría un `GROUP BY` en SQL) para comparar métricas entre grupos. Vas a resolver el análisis completo del Test A/B usando **únicamente diccionarios y listas nativas de Python, sin importar ninguna librería**.

**Tu tarea:**
1. Crea un diccionario acumulador llamado `resumen_grupos` con la estructura inicial:
   ```python
   resumen_grupos = {
       "A": {"usuarios": 0, "convertidos": 0, "tiempos": []},
       "B": {"usuarios": 0, "convertidos": 0, "tiempos": []},
   }
   ```
2. Recorre la lista `experimento_ab` con un bucle `for`:
   * Obtén el grupo de la fila (`grupo = fila["grupo"]`).
   * Suma `1` al contador `resumen_grupos[grupo]["usuarios"]`.
   * Si `fila["convirtio"]` es `True`, suma `1` a `resumen_grupos[grupo]["convertidos"]`.
   * Si `fila["tiempo_seg"]` **no es `None`**, añade ese tiempo a la lista `resumen_grupos[grupo]["tiempos"]`.
3. Crea un diccionario final `resultado_ab = {}` recorriendo `resumen_grupos.items()` para calcular en cada grupo:
   * `tasa_conv_pct`: `(convertidos / usuarios) * 100`, redondeada a 1 decimal.
   * `tiempo_medio_seg`: promedio de la lista `tiempos`, redondeado a 1 decimal.
4. Imprime el resumen comparativo del Grupo A frente al Grupo B.

> 💡 **Pista:** Accede a los valores internos encadenando corchetes, por ejemplo: `resumen_grupos[grupo]["usuarios"] += 1`.
> ✅ **Resultado esperado:** `Grupo A: 0.0% conversión (40.0s) | Grupo B: 100.0% conversión (102.5s)`

````python
# Usamos la misma tabla experimento_ab (sin importar librerías externas)
experimento_ab = [
    {"user": "U1", "grupo": "A", "tiempo_seg": 40, "convirtio": False},
    {"user": "U2", "grupo": "B", "tiempo_seg": 95, "convirtio": True},
    {"user": "U3", "grupo": "A", "tiempo_seg": None, "convirtio": False},
    {"user": "U4", "grupo": "B", "tiempo_seg": 110, "convirtio": True},
]

# Escribe aquí tu solución para el Ejercicio 2.5 usando diccionarios
# Crea un diccionario acumulador:
resumen_grupos = {
    "A": {"usuarios": 0, "convertidos": 0, "tiempos": []},
    "B": {"usuarios": 0, "convertidos": 0, "tiempos": []}
}

# Bucle For para clasificar y acumular los datos
for fila in experimento_ab:
    grupo = fila["grupo"]

    # Suma 1 al contador de usuarios del grupo
    resumen_grupos[grupo]["usuarios"] += 1

    # Si convirtió es True, suma 1 a los convertidos del grupo
    if fila["convirtio"]:
        resumen_grupos[grupo]["convertidos"] += 1

    # Si el tiempo no es None, lo agregamos a su lista de tiempos
    if fila["tiempo_seg"] is not None:
        resumen_grupos[grupo]["tiempos"].append(fila["tiempo_seg"])

# Crea un diccionario final resultado_ab = {}
resultado_ab = {}

# Bucle para calcular las métricas por cada grupo
for grupo, datos in resumen_grupos.items():
    # Calcular tasa_conv_pct
    tasa_conv_pct = round((datos["convertidos"] / datos["usuarios"]) * 100, 1) if datos["usuarios"] > 0 else 0.0

    # Calcular tiempo_medio_seg
    tiempo_medio_seg = round(sum(datos["tiempos"]) / len(datos["tiempos"]), 1) if len(datos["tiempos"]) > 0 else 0.0

    # Guardar en el diccionario de resultados
    resultado_ab[grupo] = {"tasa_conv_pct": tasa_conv_pct, "tiempo_medio_seg": tiempo_medio_seg}

# Imprime el resumen comparativo del Grupo A frente al Grupo B.
print(f"Grupo A: {resultado_ab['A']['tasa_conv_pct']}% conversión ({resultado_ab['A']['tiempo_medio_seg']}s) | Grupo B: {resultado_ab['B']['tasa_conv_pct']}% conversión ({resultado_ab['B']['tiempo_medio_seg']}s)")
````

### 🟢 Ejercicio 3.1 — Filtrado de Lecturas de un Sensor IoT *(Calentamiento)*
**Contexto:** Un sensor de temperatura en un almacén envía el código `-999` cada vez que pierde la conexión Wi-Fi. Si no lo filtramos, arruinará la temperatura media.

**Tu tarea:**
1. Crea una lista vacía llamada `lecturas_validas`.
2. Recorre `temperaturas_sensor` con un bucle `for`.
3. Si la temperatura es distinta de `-999` (`!= -999`), añádela a `lecturas_validas`.
4. Imprime cuántas lecturas válidas quedaron y cuál es la temperatura media real (redondeada a 2 decimales).

````python
# Escribe aquí tu solución para el Ejercicio 4.2

#normalizar_texto(texto) que: Si texto is None (o está vacío ""), devuelva "Sin Dato".

def normalizar_texto (texto):
    if texto is None or texto == "":
        return "Sin Dato"
    else:
        return str(texto).strip().title()

print(normalizar_texto("  barcelona "))
print(normalizar_texto("MADRID"))
print(normalizar_texto(None))

````

### 🔴 Ejercicio 4.5 — Pipeline ETL Completo y Reporte Ejecutivo *(Reto Integrador Final)*
**Contexto:** Has llegado al reto final del módulo. Vas a construir la función orquestadora `generar_reporte_etl(filas_crudas)` que conecta la limpieza con el cálculo de KPIs ejecutivos y métricas de calidad del dato.

**Tu tarea:**
1. Define `generar_reporte_etl(filas_crudas)`.
2. Llama dentro a `limpios = transformar_datos(filas_crudas)` (tu función del Ejercicio 4.4).
3. Calcula:
   * `registros_entrada`: `len(filas_crudas)`
   * `registros_validos`: `len(limpios)`
   * `tasa_calidad_pct`: `(registros_validos / registros_entrada) * 100` redondeado a 1 decimal (protegido con `if registros_entrada > 0 else 0.0`).
   * `facturacion_total`: suma de `"ingreso"` en `limpios`, redondeada a 2 decimales.
   * `facturacion_por_categoria`: diccionario con el total acumulado por cada categoría limpia.
4. Devuelve (`return`) un diccionario con todas esas métricas ejecutivas e imprímelo de forma clara.

> ✅ **Resultado esperado:**
> `{'registros_entrada': 4, 'registros_validos': 2, 'tasa_calidad_pct': 50.0, 'facturacion_total': 350.0, 'facturacion_por_categoria': {'Tech': 350.0}}`

````python
# Escribe aquí tu solución para el Ejercicio 4.5 (Reto Integrador Final)

def generar_reporte_etl(filas_crudas):
    # 1. Transformamos los datos usando la función del Ejercicio 4.4
    limpios = transformar_datos(filas_crudas)

    # 2. Calcular número de registros
    registros_entrada = len(filas_crudas)
    registros_validos = len(limpios)

    # 3. Tasa de calidad con protección divisional por cero de forma correcta en una línea
    tasa_calidad_pct = round((registros_validos / registros_entrada) * 100, 1) if registros_entrada > 0 else 0.0

    # 4. Facturación total redondeada a 2 decimales
    facturacion_total = round(sum(v["ingreso"] for v in limpios), 2)

    # 5. Agrupación y acumulado por categoría limpia
    facturacion_por_categoria = {}
    for registro in limpios:
        cat = registro["categoria"]
        facturacion_por_categoria[cat] = round(facturacion_por_categoria.get(cat, 0.0) + registro["ingreso"], 2)

    # Devuelve el diccionario de reporte completo
    return {
        "registros_entrada": registros_entrada,
        "registros_validos": registros_validos,
        "tasa_calidad_pct": tasa_calidad_pct,
        "facturacion_total": facturacion_total,
        "facturacion_por_categoria": facturacion_por_categoria
    }

# Dataset de prueba crudo para verificar
dataset_final_raw = [
    {"categoria": " tech ", "unidades": 2, "precio": 100.0},
    {"categoria": "hogar", "unidades": -1, "precio": 50.0},
    {"categoria": "TECH", "unidades": 1, "precio": 150.0},
    {"categoria": "moda", "unidades": 3, "precio": None},
]

# Generar e imprimir reporte
reporte = generar_reporte_etl(dataset_final_raw)
print(reporte)
````
